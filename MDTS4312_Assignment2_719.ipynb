```python
class KMeans:
  def __init__(self,n_clusters,max_iter):
    self.n_clusters=n_clusters
    self.max_iter=max_iter
    self.centroids=None
  def fit_predict(self,X):
    random_index=np.random.choice(range(0,X.shape[0]),self.n_clusters)
    self.centroids=X.iloc[random_index].values
    for i in range(self.max_iter):
      cluster_group=self.assign_clusters(X)
      old_centroid=self.centroids
      self.centroids=self.move_centroids(X,cluster_group)
      if np.all(old_centroid==self.centroids):
        break
    return cluster_group
  def assign_clusters(self,X):
    cluster_group=[]
    distances=[]
    for row in X.values:
      for centroid in self.centroids:
        distances.append(np.sqrt(np.dot(row-centroid,row-centroid)))
      min_distance=min(distances)
      index_position=distances.index(min_distance)
      cluster_group.append(index_position)
      distances.clear()
    return np.array(cluster_group)
  def move_centroids(self,X,cluster_group):
    new_centroids= []
    cluster_type=np.unique(cluster_group)
    for type in cluster_type:
      new_centroid=np.mean(X.iloc[cluster_group==type].values, axis=0)
      new_centroids.append(new_centroid)
    return np.array(new_centroids)
```

```python
km=KMeans(3,100)
clusters=km.fit_predict(X)
```

```python
print(clusters)
```

```python
X['clusters']=clusters
```
