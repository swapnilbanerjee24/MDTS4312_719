```python
import numpy as np
import pandas as pd
import random
import matplotlib.pyplot as plt
```

```python
class PageRank:
  def __init__(self, graph, max_iter):
    self.graph = graph
    self.nodes = list(graph.keys())
    self.num_nodes = len(self.nodes)
    self.max_iter = max_iter
    self.rank = {node: 1 / self.num_nodes for node in self.nodes}
    self.damping_factor = 0.85

  def page_ranking(self):
    for _ in range(self.max_iter):
      new_rank = {}
      for node in self.nodes:
        rank = (1 - self.damping_factor) / self.num_nodes
        for pr_node in self.nodes:
          if node in self.graph[pr_node]:
            rank += self.damping_factor * self.rank[pr_node] / len(self.graph[pr_node])
        new_rank[node] = rank
      self.rank = new_rank
    return self.rank



```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['B', 'E'],
    'D': ['A'],
    'E': ['D']
}

print("Example Graph:")
for node, links in graph.items():
    print(f"{node} -> {', '.join(links)}")
```

```python
pg = PageRank(graph, 10)
pg.page_ranking()
```
